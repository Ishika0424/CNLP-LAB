 5.3. Implement Part-of-Speech (POS) tagging on a given text and display the unique token, POS tag, and description of each POS tag.
                            
                   "The young student is reading an interesting book in the library."   
 (Implement two separate codes: a. Using Spacy and b.  NLTK )


```python
#5.3(a) Using spaCy

!pip install spacy
!python -m spacy download en_core_web_sm
```

```python
import spacy
from spacy import explain

# Load English model
nlp = spacy.load("en_core_web_sm")

text = "The young student is reading an interesting book in the library."

doc = nlp(text)

print("Token\t\tPOS Tag\t\tDescription")
print("-" * 65)

seen = set()

for token in doc:

    # Skip punctuation
    if token.is_punct:
        continue

    if token.text.lower() not in seen:

        seen.add(token.text.lower())

        description = explain(token.pos_)

        print(f"{token.text:15} {token.pos_:15} {description}")
```

```python
import nltk
import json

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger_eng")
nltk.download("tagsets_json")

text = "The young student is reading an interesting book in the library."

# Tokenization
tokens = nltk.word_tokenize(text)

# POS tagging
tagged_words = nltk.pos_tag(tokens)

# Get POS tag descriptions
tagset = json.loads(nltk.data.load(
    "help/tagsets_json/PY3_json/upenn_tagset.json",
    format="raw"
))

print("Token\t\tPOS Tag\t\tDescription")
print("-" * 75)

seen = set()

for word, tag in tagged_words:

    if not word.isalpha():
        continue

    if word.lower() not in seen:

        seen.add(word.lower())

        if tag in tagset:
            description = tagset[tag][0]
        else:
            description = "Unknown"

        print(f"{word:15} {tag:10} {description}")
```
