```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("employee_information_100.csv")

print(df.head())
```

# Experiment 1.1: Average salary of each department using a bar chart

```python
avg_salary = df.groupby("Department")["Salary"].mean()

print(avg_salary)

avg_salary.plot(kind="bar")
plt.title("Average Salary of Each Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Experiment 1.2: Number of employees working in each department

```python
employee_count = df["Department"].value_counts()

print("Number of Employees in Each Department:")
print(employee_count)


employee_count.plot(kind="bar")

plt.title("Number of Employees in Each Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Experiment 1.3: Percentage of male and female employees using a pie chart

```python
gender_count = df["Gender"].value_counts()

gender_count.plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Percentage of Male and Female Employees")
plt.ylabel("")
plt.show()
```

# Experiment 1.4: Histogram showing salary distribution

```python
plt.hist(df["Salary"], bins=10, edgecolor="black")

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")
plt.show()
```

# Experiment 1.5: Scatter plot showing relationship between experience and salary

```python
plt.scatter(df["Experience_Years"], df["Salary"])

plt.title("Experience vs Salary")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.show()
```

# Experiment 1.6: Top 10 employees with the highest salary

```python
top_10 = df.nlargest(10, "Salary")

print("Top 10 Employees with Highest Salary:")
print(top_10)

print(top_10[["Name", "Salary"]])
```

# Experiment 1.7: Highest salary in every department

```python
highest_salary = df.groupby("Department")["Salary"].max()

print("Highest Salary in Every Department:")
print(highest_salary)
```

# Experiment 1.8: Employees whose salary is greater than the overall average salary

```python
overall_average = df["Salary"].mean()

print("Overall Average Salary:", overall_average)

above_average = df[df["Salary"] > overall_average]

print("\nEmployees Earning More Than Average:")
print(above_average)
```

# Experiment 1.9: Average years of experience in each department

```python
avg_experience = df.groupby("Department")["Experience_Years"].mean()

print("Average Experience in Each Department:")
print(avg_experience)
```

# Experiment 1.10: Distribution of employee ages using a histogram

```python
plt.hist(df["Age"], bins=10, edgecolor="black")

plt.title("Employee Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.show()
```
