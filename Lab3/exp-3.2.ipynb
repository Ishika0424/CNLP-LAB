
Experiment 3.2:  Write a Python program to perform lemmatization on a given input words:

cats,	dogs,   running,  runs,	ran, studies, studying,	better, children, mice,	went, ate,	
leaves,	caring,


```python
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

words = [
    "cats", "dogs", "running", "runs", "ran",
    "studies", "studying", "better", "children",
    "mice", "went", "ate", "leaves", "caring"
]

pos_tags = {
    "cats": "n",
    "dogs": "n",
    "running": "v",
    "runs": "v",
    "ran": "v",
    "studies": "v",
    "studying": "v",
    "better": "a",
    "children": "n",
    "mice": "n",
    "went": "v",
    "ate": "v",
    "leaves": "n",
    "caring": "v"
}

print("Original Word\tLemma")
print("-" * 30)

for word in words:
    lemma = lemmatizer.lemmatize(word, pos=pos_tags[word])
    print(f"{word:<15}{lemma}")
```
