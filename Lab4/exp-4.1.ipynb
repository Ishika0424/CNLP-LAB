Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
Identify the top 10 most frequent terms and display their corresponding frequencies.

```python
import spacy
import csv
from collections import Counter

# Load spaCy English model
nlp = spacy.load("en_core_web_sm")

# Read text from file
with open("input.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Process text using spaCy
doc = nlp(text)

# Extract words
words = []

for token in doc:
    if token.is_alpha:
        words.append(token.text.lower())

# Calculate term frequency
frequency = Counter(words)

# Save frequency in CSV file
with open("term_frequency.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(["Term", "Frequency"])

    for term, freq in frequency.items():
        writer.writerow([term, freq])

print("Term frequencies saved in term_frequency.csv")

# Display top 10 terms
print("\nTop 10 Most Frequent Terms:")

for term, freq in frequency.most_common(10):
    print(term, ":", freq)

# Named Entity Recognition
print("\nNamed Entities:")

for entity in doc.ents:
    print(entity.text, "->", entity.label_)
```
