4.3. Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:
Read multiple text documents.
Convert text to lowercase.
Remove punctuation.
Tokenize the documents.
Calculate TF for every term.
Calculate DF for every term.
Calculate IDF using the logarithmic formula.
Calculate TF-IDF for every term in every document.
Display Term, TF, DF, IDF, and TF-IDF at each step.
Identify the top 10 terms with the highest TF-IDF score for each document.
Given Input
Consider the following three documents:

Document 1:
Natural language processing is a field of artificial intelligence.

Document 2:
Natural language processing helps computers understand human language.

Document 3:
Machine learning is an important part of artificial intelligence.


```python
import math
import string

documents = [
    "Natural language processing is a field of artificial intelligence.",
    "Natural language processing helps computers understand human language.",
    "Machine learning is an important part of artificial intelligence."
]

# -------------------------
# PREPROCESSING
# -------------------------

processed_documents = []

for document in documents:

    # Lowercase
    document = document.lower()

    # Remove punctuation
    for p in string.punctuation:
        document = document.replace(p, "")

    # Tokenization
    words = document.split()

    processed_documents.append(words)


print("Processed Documents:")

for i, words in enumerate(processed_documents):
    print("Document", i + 1, ":", words)


# -------------------------
# CALCULATE DF
# -------------------------

df = {}

for words in processed_documents:

    unique_words = set(words)

    for word in unique_words:
        if word in df:
            df[word] += 1
        else:
            df[word] = 1


# Number of documents
N = len(documents)


# -------------------------
# CALCULATE IDF
# -------------------------

idf = {}

for word in df:
    idf[word] = math.log(N / df[word])


# -------------------------
# CALCULATE TF AND TF-IDF
# -------------------------

for i, words in enumerate(processed_documents):

    print("\n--------------------------")
    print("Document", i + 1)
    print("--------------------------")

    tf = {}

    # Calculate TF
    for word in words:
        if word in tf:
            tf[word] += 1
        else:
            tf[word] = 1

    # Divide frequency by total number of words
    total_words = len(words)

    for word in tf:
        tf[word] = tf[word] / total_words

    tfidf = {}

    print("Term\tTF\tDF\tIDF\tTF-IDF")

    for word in tf:

        score = tf[word] * idf[word]

        tfidf[word] = score

        print(
            word,
            round(tf[word], 4),
            df[word],
            round(idf[word], 4),
            round(score, 4),
            sep="\t"
        )

    # Sort TF-IDF
    sorted_terms = sorted(
        tfidf.items(),
        key=lambda x: x[1],
        reverse=True
    )

    print("\nTop 10 Terms:")

    for word, score in sorted_terms[:10]:
        print(word, ":", round(score, 4))
```
