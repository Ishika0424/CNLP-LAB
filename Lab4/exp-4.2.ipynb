4.2. To perform Term-Frequency Analysis on a given text without using any toolkit.

Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
Identify the top 10 most frequent terms and display their corresponding frequencies.


```python
import string
import csv

with open("input.txt", "r", encoding="utf-8") as file:
    text = file.read()

text = text.lower()

for p in string.punctuation:
    text = text.replace(p, " ")

words = text.split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

with open("term_frequency.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(["Term", "Frequency"])

    for term, freq in frequency.items():
        writer.writerow([term, freq])

print("Term frequencies saved in term_frequency.csv")

sorted_frequency = sorted(
    frequency.items(),
    key=lambda x: x[1],
    reverse=True
)

print("\nTop 10 Most Frequent Terms:")

for term, freq in sorted_frequency[:10]:
    print(term, ":", freq)
```
