# Experiment 2.1 

```python
import string
import re
```

```python
# Read text from file
file = open("2.1_text_data.txt", "r")
text = file.read()
file.close()

print("Original Text:")
print(text)
```

# 1. Convert text to lowercase. (Also display how many uppercase?)


```python
uppercase_count = sum(1 for ch in text if ch.isupper())
text = text.lower()

print("\nNumber of uppercase letters:", uppercase_count)
```

# 2. Remove punctuation.(Also display how many different types of punctuations?)

```python
punctuations = set(ch for ch in text if ch in string.punctuation)

print("Different types of punctuation:", punctuations)
print("Number of different punctuation types:", len(punctuations))

text = text.translate(str.maketrans("", "", string.punctuation))
```

# 3. Remove numbers.(Also display how many numbers?)

```python
number_count = sum(1 for ch in text if ch.isdigit())

print("Number of digits:", number_count)

text = re.sub(r'\d+', '', text)
```

# 4. Remove extra whitespace. (How much extra whitespace is present?)

```python
spaces = re.findall(r'\s+', text)
extra_whitespace = sum(len(space) - 1 for space in spaces)

print("Extra whitespace present:", extra_whitespace)

text = ' '.join(text.split())
```

# 5. Display the cleaned text.

```python
print("\nCleaned Text:")
print(text)
```
