# 2.3) Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately)

```python
file = open("2.3_clean_data.txt", "r")
text = file.read()
file.close()

stop_words = ["is", "the", "a", "an", "in", "of", "and", "to", "on", "for"]

tokens = text.lower().split()

filtered_tokens = []  #remove

for word in tokens:
    if word not in stop_words:
        filtered_tokens.append(word)

print("Original Tokens:")
print(tokens)

print("\nStop Words:")
print(stop_words)

print("\nFiltered Tokens:")
print(filtered_tokens)
```
