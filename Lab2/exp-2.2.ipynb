# 2.2: Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately. (Three separate code: using NLTK, using spaCy and without using NLTK/spacy)

```python
# 1. Using NLTK
# !pip install nltk
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

file = open("2.2_tokenization_data.txt", "r")
text = file.read()
file.close()

words = word_tokenize(text)
sentences = sent_tokenize(text)

print("Word Tokens:")
print(words)

print("\nSentence Tokens:")
print(sentences)
```

```python
# 2. Using spaCy

# !pip install spacy
# !python -m spacy download en_core_web_sm

import spacy

nlp = spacy.load("en_core_web_sm")

# Read text
file = open("2.2_tokenization_data.txt", "r")
text = file.read()
file.close()

# Process the text
doc = nlp(text)

words = [token.text for token in doc]

sentences = [sentence.text for sentence in doc.sents]

#results
print("Word Tokens:")
print(words)

print("\nSentence Tokens:")
print(sentences)
```

```python
#3. Without NLTK or spaCy
# Read text
file = open("2.2_tokenization_data.txt", "r")
text = file.read()
file.close()

words = text.split()

sentences = text.split(".")
#results
print("Word Tokens:")
print(words)

print("\nSentence Tokens:")
print(sentences)
```
