 5.2. Implement subword-level tokenization using a SentencePiece tokenizer. Given same input data of 5.1 

Tokenize the sentence into subword units.
Display the generated tokens and their corresponding token IDs.
    (Implement two separate codes: a. Using pretrained model and b.  Without using pretrained model)

```python
#5.2(a) Using a pretrained SentencePiece model

from transformers import T5Tokenizer

# Load pretrained SentencePiece tokenizer
tokenizer = T5Tokenizer.from_pretrained("t5-small")

# Input sentence
text = "Natural language processing is a field of artificial intelligence."
print("Input:", text)

# Tokenize
tokens = tokenizer.tokenize(text)

# Convert tokens into IDs
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("\nToken\t\tToken ID")
print("-" * 35)

for token, token_id in zip(tokens, token_ids):
    print(f"{token:20} {token_id}")
```

```python
# 5.2(b) SentencePiece without pretrained model

import sentencepiece as spm

# Train SentencePiece model using our own corpus
spm.SentencePieceTrainer.train(
    input="input.txt",
    model_prefix="my_tokenizer",
    vocab_size=300,
    model_type="bpe"
)

# Load our trained model
sp = spm.SentencePieceProcessor()

sp.load("my_tokenizer.model")

# Input sentence
text = "Natural language processing is a field of artificial intelligence."
print("Input:", text)

# Generate subword tokens
tokens = sp.encode(text, out_type=str)

# Generate token IDs
token_ids = sp.encode(text, out_type=int)

print("\nToken\t\tToken ID")
print("-" * 35)

for token, token_id in zip(tokens, token_ids):
    print(f"{token:20} {token_id}")
```
