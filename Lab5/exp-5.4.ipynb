  5.4. Implement Part-of-Speech (POS) tagging on a given text and display the unique token, POS tag, description of each token and frequency using Spacy.
            Input data: Same input data of 5.1


```python
import spacy
from collections import Counter
from spacy import explain

# Load spaCy model
nlp = spacy.load("en_core_web_sm")

# Read input file
with open("input.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Process text
doc = nlp(text)

# Store words only
words = [
    token.text.lower()
    for token in doc
    if not token.is_punct and not token.is_space
]

# Calculate frequency
frequency = Counter(words)

print("Token\t\tPOS Tag\t\tDescription\t\tFrequency")
print("-" * 85)

seen = set()

for token in doc:

    if token.is_punct or token.is_space:
        continue

    word = token.text.lower()

    # Display each token only once
    if word not in seen:

        seen.add(word)

        pos = token.pos_
        description = explain(pos)
        freq = frequency[word]

        print(
            f"{word:20} "
            f"{pos:15} "
            f"{str(description):30} "
            f"{freq}"
        )
```
